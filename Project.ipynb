```python
import pandas as pd
from pathlib import Path
from typing import List, Union

def export_tables(
    tables: Union[pd.DataFrame, List[pd.DataFrame]],
    output_path: str,
    file_format: str = "xlsx"  # "xlsx" or "csv"
) -> None:
    """
    Export one or more tables to CSV or Excel.

    Parameters
    ----------
    tables : DataFrame or list of DataFrames
        The extracted & cleaned table(s)
    output_path : str
        Desired output file path (extension will be adjusted if needed)
    file_format : str
        "xlsx" or "csv" (default: "xlsx")
    """

    # Normalize input to a list
    if isinstance(tables, pd.DataFrame):
        tables = [tables]
    
    if not tables:
        print("No tables to export.")
        return

    output_path = Path(output_path)
    file_format = file_format.lower().strip(".")

    # ---------- Excel Export ----------
    if file_format in ["xlsx", "excel"]:
        # Force correct extension
        if output_path.suffix.lower() != ".xlsx":
            output_path = output_path.with_suffix(".xlsx")

        with pd.ExcelWriter(output_path, engine="openpyxl") as writer:
            for i, df in enumerate(tables, start=1):
                sheet_name = f"Table_{i}"
                # Excel sheet name limit is 31 characters
                sheet_name = sheet_name[:31]
                df.to_excel(writer, sheet_name=sheet_name, index=False)

        print(f"Successfully exported {len(tables)} table(s) to → {output_path}")

    # ---------- CSV Export ----------
    elif file_format == "csv":
        if len(tables) == 1:
            # Single table → one CSV file
            if output_path.suffix.lower() != ".csv":
                output_path = output_path.with_suffix(".csv")
            tables[0].to_csv(output_path, index=False, encoding="utf-8-sig")
            print(f"Successfully exported table to → {output_path}")
        else:
            # Multiple tables → create numbered CSV files
            output_dir = output_path.parent / output_path.stem
            output_dir.mkdir(parents=True, exist_ok=True)

            for i, df in enumerate(tables, start=1):
                csv_path = output_dir / f"table_{i}.csv"
                df.to_csv(csv_path, index=False, encoding="utf-8-sig")

            print(f"Successfully exported {len(tables)} tables to folder → {output_dir}")

    else:
        raise ValueError("file_format must be either 'xlsx' or 'csv'")
```
